```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import rasterio as rio
import rasterio.plot as rplt
import cmocean as cmo
import app
```

```python
dmg_path = r"D:\PhD\21_Experiments\TidesDamageDriver\03_cleaned\dmg_merged\2020"
img_base = r"S1_20201101_20201110_30m_output_10px"
```

```python
img_name_dmg = os.path.join(dmg_path, img_base + "_dmg.tif")
img_name_crev = os.path.join(dmg_path, img_base + "_crevSig.tif")
img_name_alphaC = os.path.join(dmg_path, img_base + "_alphaC.tif")
img_name_deltaalpha = os.path.join(dmg_path, img_base + "_delta-alpha.tif")
img_name_emax = os.path.join(dmg_path, img_base + "_emax.tif")
img_name_act = os.path.join(dmg_path, img_base + "_act.tif")
```

```python
tiles = []
transform = []

for img_name in [img_name_dmg, img_name_emax, img_name_act]:
    with rio.open(img_name) as src:
        tiles.append(src.read(1))
        transform.append(src.transform)
```

```python
fig, ax = plt.subplots(1, 4, figsize=(16, 5), sharex=True, sharey=True)
rplt.show(tiles[0], ax=ax[0], cmap="cmo.matter", transform=transform[0])
rplt.show(tiles[1], ax=ax[1], cmap="cmo.tempo", transform=transform[1])
rplt.show(tiles[2], ax=ax[2], cmap="cmo.rain_r", transform=transform[2])
rplt.show(tiles[2], ax=ax[3], cmap="cmo.rain_r", transform=transform[2])
plt.show()
```
