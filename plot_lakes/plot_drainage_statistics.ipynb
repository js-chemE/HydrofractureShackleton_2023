```python
import app
import os
import matplotlib.pyplot as plt
import rasterio.plot as rplt
import rasterio
import numpy as np
import geopandas as gpd

app.setup_logger(use_console_handler=True, use_file_handler=True)
```

```python
all_years = ["2016", "2018", "2019", "2020"] # , "2018", "2019", "2020", "2022", "2023"
all_lakes = []
for year in all_years:
    drainage_80_path = os.path.join(r"D:\PhD\21_Experiments\TidesDamageDriver\02_processed\01_drainages_v2", year)
    drainage_50_path = os.path.join(r"D:\PhD\21_Experiments\TidesDamageDriver\02_processed\01_drainages_50", year)
    drainage_70_path = os.path.join(r"D:\PhD\21_Experiments\TidesDamageDriver\02_processed\01_drainages_70", year)
    lakes_80 = gpd.read_file(os.path.join(drainage_80_path, f"{year}_drainages_0.shp"))
    if os.path.exists(drainage_50_path):
        lakes_50 = gpd.read_file(os.path.join(drainage_50_path, f"{year}_drainages_0.shp"))
    if os.path.exists(drainage_50_path):
        lakes_70 = gpd.read_file(os.path.join(drainage_70_path, f"{year}_drainages_0.shp"))
    all_lakes.append(lakes_80)
    print(f"{lakes_50["criteria"].unique().size:03d} - {lakes_70["criteria"].unique().size:03d} - {lakes_80["criteria"].unique().size:03d} lakes in {year}")
```
