```python
from shapely.geometry import Point
events_raw = pd.read_excel(path, index_col=0).query("comment != 'refreeze'")
events_raw["geometry"] = [Point(xy) for xy in zip(events_raw["lon"], events_raw["lat"])]

#events_raw["geometry"] = gpd.GeoSeries.from_wkt(events_raw["geom_lake0_3031"])
events = gpd.GeoDataFrame(events_raw, geometry="geometry")  # type: ignore

events.set_crs("EPSG:4326", inplace=True)
events.sort_values("lon", inplace=True)
events.reset_index(drop=True, inplace=True)
events.to_crs("EPSG:3031", inplace=True)
events["geometry"] = events["geometry"].buffer(10)
events["label"] = list(string.ascii_uppercase)[: len(events)]
events["year"] = [app.utils.dates.determine_antarctic_summer(d) for d in events["date-0 vis"]]
events["volume-1 m^3"] = np.nan_to_num(x=events["volume-1 m^3"].values, nan=0)  # type: ignore
#events[["year", "label", "geometry"]].to_file(os.path.join(data_folder, "shackleton_hydrofractures.shp"))

subset = events.loc[events.index.isin(to_added)]
subset.reset_index(drop=True, inplace=True)
subset
```

```python
from shapely import wkt

subset_4326 = subset.to_crs(epsg=4326, inplace=False)
subset_4326["centroid"] = [wkt.loads(g).centroid for g in subset_4326["geom_lake0_3031"]]
subset_4326
```

```python
# Rename columns to align
df2_aligned = subset.rename(columns={
    "lake": "lake id",
    "volume-0 m^3": "volume-0",
    "volume-1 m^3": "volume-1",
    "meandepth-0 m": "mean-0",
    "meandepth-1 m": "mean-1",
    #"geometry": "geometry_point",
    #"geom_lake0_3031" : "geometry"
})

# Keep only matching columns
matching_cols = [col for col in gdf_0.columns if col in df2_aligned.columns]
df2_subset = df2_aligned[matching_cols].copy()
# Fill in missing columns with NA
for col in gdf_0.columns:
    if col not in df2_subset.columns:
        df2_subset[col] = pd.NA

# Ensure same CRS
if gdf_0.crs != df2_subset.crs:
    df2_subset = df2_subset.to_crs(gdf_0.crs)

# Order columns
df2_subset = df2_subset[gdf_0.columns]

print("Columns in gdf_0:", gdf_0.columns)
print("Columns in df2_subset:", df2_subset.columns)
# Concatenate
df_combined = pd.concat([gdf_0, df2_subset], ignore_index=True)
```
